### 1. Data Loading and Imports
In this section, we import the necessary libraries and load the Netflix dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://huggingface.co/datasets/hugginglearners/netflix-shows/resolve/main/netflix_titles.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

### 3. Visualizing Content Type Distribution
We compare the proportion of Movies versus TV Shows available on Netflix.

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='type')

plt.title("Distribution of Movies and TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Count")

plt.show()
```

### 4. Geographical Content Distribution
Visualizing the top countries that produce the most Netflix titles.

```python
top_countries = df['country'].dropna().str.split(', ').explode().value_counts().head(10)

plt.figure(figsize=(10, 5))

sns.barplot(x=top_countries.values, y=top_countries.index)

plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

### 5. Historical Trends & Release Years
Let's trace how many shows/movies were released over the years.

```python
year_counts = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(12, 5))

plt.plot(year_counts.index, year_counts.values)

plt.title("Netflix Content Release Trends")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

### 6. Temporal Analysis of Content Additions
Analyzing when content was added to the platform by extracting year information.

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')

df['year_added'] = df['date_added'].dt.year

df[['date_added', 'year_added']].head()
```

```python
correlation = df[['release_year', 'year_added']].corr()

print(correlation)
```

```python
plt.figure(figsize=(6, 4))

sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap of Netflix Dataset")

plt.show()
```

```python
content_trend = df.groupby(['year_added', 'type']).size().reset_index(name='count')

plt.figure(figsize=(12, 5))

sns.lineplot(data=content_trend, x='year_added', y='count', hue='type')

plt.title("Movies and TV Shows Added to Netflix Over the Years")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")

plt.show()
```

### 7. Content Rating & Demographics
Looking at the distribution of content ratings across titles.

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    y='rating',
    order=df['rating'].value_counts().head(10).index
)

plt.title("Top 10 Netflix Content Ratings")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")

plt.show()
```

### 8. Movie Duration Distribution
Examining the runtime (in minutes) of movie titles.

```python
movies = df[df['type'] == 'Movie'].copy()

movies['duration_minutes'] = pd.to_numeric(
    movies['duration'].str.replace(' min', '', regex=False),
    errors='coerce'
)

print(movies['duration_minutes'].describe())
```

### 9. Genre and Classification Analysis
Splitting and identifying the most common genres in which Netflix titles are listed.

```python
genres = df['listed_in'].dropna().str.split(', ').explode()

top_genres = genres.value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(x=top_genres.values, y=top_genres.index)

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.show()
```
